```python
import os

from IPython.display import Audio, display
from ipywidgets import FileUpload
from huggingface_hub import InferenceClient
```

**B1:** Tải âm thanh lên

```python
uploader = FileUpload(accept='.wav,.flac,.mp3,.onnx,.json', multiple=True)

display(uploader)
```

```python
for index in range(0,len(uploader.value)):
    uploaded = uploader.value[index]
    
    filename = uploaded["name"]
    content = uploaded["content"]
    # A Sang thay đường dẫn tới vị trí muốn lưu
    output_path = os.path.join("/kaggle/working", filename)
    
    with open(output_path, "wb") as f:
        f.write(content.tobytes())
    
    print("Saved to:", output_path)
```

```python
os.system("pip install -q piper-tts")

from piper import PiperVoice
import wave

model_path = "/kaggle/input/datasets/tunnvq/videoaiset/ngochuyennew.onnx"
config_path = "/kaggle/input/datasets/tunnvq/videoaiset/ngochuyennew.onnx.json"

voice = PiperVoice.load(model_path, config_path=config_path)

output_path = "/kaggle/working/output_piper.wav"

with wave.open(output_path, "wb") as wav_file:
    wav_file.setnchannels(1)
    wav_file.setsampwidth(2)
    wav_file.setframerate(voice.config.sample_rate)

    for audio_chunk in voice.synthesize(vietnamese_text):
        wav_file.writeframes(audio_chunk.audio_int16_bytes)

print(f"Generated speech saved to {output_path}")
display(Audio(output_path))
```

**Hướng 2:** Nhái giọng gốc.

```python
# --- Step 1: Install dependencies ---
os.system(
    "pip install -q torch==2.8.0+cu128 torchaudio==2.8.0+cu128 torchvision==0.23.0+cu128 "
    "--extra-index-url https://download.pytorch.org/whl/cu128"
)
os.system("pip install -q omnivoice")
import torch
import torchaudio
from omnivoice import OmniVoice
import numpy as np

# --- Step 2: Load the Vietnamese fine-tuned model ---
device = "cuda:0" if torch.cuda.is_available() else "cpu"
print(device)

model = OmniVoice.from_pretrained(
    "splendor1811/omnivoice-vietnamese",
    device_map=device,
    dtype=torch.float16,
)

# --- Step 3: Zero-shot voice cloning from reference audio ---
reference_wav = mp3_path

max_ref_seconds = 10

waveform, sr = torchaudio.load(reference_wav)
max_samples = int(max_ref_seconds * sr)

if waveform.shape[1] > max_samples:
    waveform = waveform[:, :max_samples]

trimmed_ref_path = "/kaggle/working/reference_trimmed.wav"
torchaudio.save(trimmed_ref_path, waveform, sr)

reference_wav = trimmed_ref_path
print(f"Trimmed reference audio saved to {reference_wav} ({waveform.shape[1]/sr:.1f}s)")

asr_output = client.automatic_speech_recognition(trimmed_ref_path, model="Qwen/Qwen3-ASR-0.6B")
reference_text = asr_output.text

output_path = "/kaggle/working/output_vi.wav"

# Model tự tạo prompt, nhanh hơn
voice_prompt = model.create_voice_clone_prompt(
    ref_audio=reference_wav,
    ref_text=reference_text,
)
audio = model.generate(
    text=vietnamese_text,
    language="vietnamese",
    voice_clone_prompt=voice_prompt,
)

# Model dùng trực tiếp, chậm hơn
# audio = model.generate(
#     text=vietnamese_text,
#     language="vietnamese",
#     ref_audio=reference_wav,
#     ref_text=reference_text,
# )

wav = audio[0]

if isinstance(wav, np.ndarray):
    wav = torch.from_numpy(wav)

if wav.dim() == 1:
    wav = wav.unsqueeze(0)

wav = wav.to(torch.float32)

torchaudio.save(output_path, wav, 24000)
print(f"Generated speech saved to {output_path}")
display(Audio(output_path))
```
